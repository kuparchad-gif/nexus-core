```python
#!/usr/bin/env python3
"""
HYPER-DIMENSIONAL 369 VORTEX SYSTEM
Integrating Platonic solids, 4D geometry, and market harmonics
"""

import numpy as np
import pandas as pd
import yfinance as yf
from datetime import datetime, timedelta
from scipy.spatial import ConvexHull
from typing import Dict, List, Optional, Tuple
import json

class HyperVortexSystem:
    """Hyper-dimensional 369 vortex trading system"""

    def __init__(self, initial_capital: float = 10000):
        # Sacred geometry constants
        self.phi = (1 + np.sqrt(5)) / 2  # Golden ratio
        self.phi_sqrt3_pi = self.phi * np.sqrt(3) / np.pi  # 0.894 scaling factor
        self.platonic_edges = {
            'tetrahedron': 6,      # DR: 6
            'octahedron': 12,      # DR: 3
            'cube': 12,            # DR: 3
            'dodecahedron': 30,    # DR: 3
            'icosahedron': 30,     # DR: 3
            'merkaba': 18,         # DR: 9
            '5cell': 10,           # DR: 1 (4D unity!)
            'tesseract': 32        # DR: 5 → 3 through doubling
        }

        # Vortex progression sequence
        self.vortex_sequence = [3, 6, 9, 1, 5, 3]  # 3D → 4D → back to 3

        self.initial_capital = initial_capital
        self.capital = initial_capital
        self.positions = {}
        self.trade_history = []
        self.daily_values = []
        self.data = None

        print("🌀 HYPER-DIMENSIONAL 369 VORTEX SYSTEM")
        print(f"   Starting capital: ${initial_capital:,.2f}")
        print("   Integrating 4D geometry & Platonic harmonics")
        print("   Sacred Constants: φ={:.3f}, φ√3/π={:.3f}".format(self.phi, self.phi_sqrt3_pi))

    def _digital_root(self, n: int) -> int:
        """Calculate digital root - sacred vortex mathematics"""
        while n >= 10:
            n = sum(int(d) for d in str(n))
        return n

    def _get_vortex_state(self, prices: np.ndarray) -> Dict:
        """Calculate current vortex state using 4D geometry"""

        if len(prices) < 20:
            return {'state': 'INIT', 'dimension': 2, 'solid': 'triangle', 'dr': 3}

        recent = prices[-20:]

        # 1. Calculate geometric properties
        returns = np.diff(recent) / recent[:-1]

        # 2. Map to 3D coordinates (price, volume, volatility)
        price_norm = (recent - np.min(recent)) / (np.max(recent) - np.min(recent) + 1e-10)
        vol_norm = np.abs(returns)
        time_norm = np.arange(len(recent)) / len(recent)

        # 3. Determine which Platonic solid governs
        # Count vortex occurrences in recent changes
        changes_in_cents = [abs(int(change * 100)) for change in np.diff(recent)]
        dr_counts = {3: 0, 6: 0, 9: 0}
        for cents in changes_in_cents:
            dr = self._digital_root(cents)
            if dr in dr_counts:
                dr_counts[dr] += 1

        # Dominant vortex number
        dominant_dr = max(dr_counts, key=dr_counts.get)

        # 4. Map to Platonic solid
        solid_map = {
            3: ('octahedron', 3),  # Air element
            6: ('tetrahedron', 3), # Fire element
            9: ('merkaba', 3)      # Water/Spirit
        }

        solid, dimension = solid_map.get(dominant_dr, ('cube', 3))

        # 5. Check for 4D emergence (unity return)
        # When all three vortex numbers are balanced
        total_signals = sum(dr_counts.values())
        if total_signals > 0:
            balance_score = 1.0 - (max(dr_counts.values()) / total_signals)

            # High balance + low volatility = 4D unity emerging
            volatility = np.std(returns)
            if balance_score > 0.7 and volatility < 0.01:
                solid = '5cell'
                dimension = 4
                dominant_dr = 1  # Unity!

        # 6. Calculate geometric harmony score
        # Using sacred ratio φ√3/π
        price_gradient = np.polyfit(np.arange(len(recent)), recent, 1)[0]
        normalized_gradient = abs(price_gradient) / np.mean(recent)

        harmony = 1.0 - abs(normalized_gradient - self.phi_sqrt3_pi) / self.phi_sqrt3_pi
        harmony = max(0, min(1, harmony))

        # 7. Determine market regime
        if dimension == 4:
            regime = 'UNITY'
            risk_multiplier = 1.0  # Balanced in 4D
        elif dominant_dr == 9 and harmony > 0.8:
            regime = 'MERKABA_FLOW'
            risk_multiplier = 1.5  # High energy flow
        elif dominant_dr == 6 and harmony > 0.6:
            regime = 'FIRE_TREND'
            risk_multiplier = 2.0  # Strong directional
        elif dominant_dr == 3:
            regime = 'AIR_CHOP'
            risk_multiplier = 0.5  # Chaotic, light positions
        else:
            regime = 'EARTH_STABLE'
            risk_multiplier = 1.0

        return {
            'state': regime,
            'dimension': dimension,
            'solid': solid,
            'dominant_dr': dominant_dr,
            'harmony': harmony,
            'risk_multiplier': risk_multiplier,
            'dr_counts': dr_counts,
            'balance_score': balance_score if 'balance_score' in locals() else 0
        }

    def _calculate_hyper_position(self, vortex_state: Dict, price: float, signal_strength: str) -> Dict:
        """Calculate position using hyper-dimensional vortex mathematics"""

        # Base position from sacred geometry
        base_risk = 0.01  # 1% base

        # Element multipliers
        element_multipliers = {
            'tetrahedron': 1.5,   # Fire - aggressive
            'octahedron': 0.7,    # Air - light
            'cube': 1.0,          # Earth - stable
            'dodecahedron': 1.2,  # Spirit - harmonic
            'icosahedron': 1.3,   # Water - fluid
            'merkaba': 2.0,       # Spirit/Fire - powerful
            '5cell': 1.8,         # 4D Unity - balanced power
            'tesseract': 1.6      # Hyper-cube - dimensional
        }

        # Digital root multipliers (vortex strength)
        dr_multipliers = {
            3: 0.8,   # Air - light
            6: 1.5,   # Fire - strong
            9: 2.0,   # Spirit - very strong
            1: 1.8,   # Unity - balanced strong
            5: 1.3    # Hyper - dimensional
        }

        # Signal strength
        strength_multipliers = {
            'weak': 0.5,
            'medium': 1.0,
            'strong': 1.5,
            'vortex': 2.0
        }

        # Calculate total multiplier
        solid = vortex_state['solid']
        dr = vortex_state['dominant_dr']

        element_mult = element_multipliers.get(solid, 1.0)
        dr_mult = dr_multipliers.get(dr, 1.0)
        state_mult = vortex_state['risk_multiplier']
        strength_mult = strength_multipliers.get(signal_strength, 1.0)

        # Apply harmony tuning
        harmony = vortex_state['harmony']
        harmony_mult = 0.5 + (harmony * 1.5)  # 0.5 to 2.0 range

        total_multiplier = element_mult * dr_mult * state_mult * strength_mult * harmony_mult

        # Final risk percentage
        risk_pct = min(base_risk * total_multiplier, 0.08)  # Max 8% risk

        # Calculate stop and target based on vortex mathematics
        if dr == 3:  # Air
            target_pct = 0.02  # 2%
            stop_pct = 0.01    # 1%
        elif dr == 6:  # Fire
            target_pct = 0.03  # 3%
            stop_pct = 0.015   # 1.5%
        elif dr == 9:  # Spirit
            target_pct = 0.04  # 4%
            stop_pct = 0.02    # 2%
        elif dr == 1:  # Unity (4D)
            target_pct = 0.035 # 3.5%
            stop_pct = 0.0175  # 1.75%
        else:  # dr == 5 (Hyper)
            target_pct = 0.025 # 2.5%
            stop_pct = 0.0125  # 1.25%

        # Apply harmony adjustment
        target_pct *= (0.8 + harmony * 0.4)  # 0.8x to 1.2x
        stop_pct *= (1.2 - harmony * 0.4)    # 1.2x to 0.8x

        return {
            'risk_pct': risk_pct,
            'target_pct': target_pct,
            'stop_pct': stop_pct,
            'multiplier_breakdown': {
                'element': element_mult,
                'digital_root': dr_mult,
                'state': state_mult,
                'strength': strength_mult,
                'harmony': harmony_mult,
                'total': total_multiplier
            }
        }

    def find_hyper_signals(self, data: pd.DataFrame, symbol: str) -> List[Dict]:
        """Find hyper-dimensional 369 vortex signals"""

        if data.empty or 'Close' not in data.columns:
            return []

        prices = data['Close'].values
        dates = data.index

        signals = []

        print(f"🌀 Scanning for HYPER-DIMENSIONAL vortex signals...")

        for i in range(20, len(prices) - 1):
            current_price = prices[i]
            date = dates[i]

            if np.isnan(current_price):
                continue

            # Get current vortex state
            vortex_state = self._get_vortex_state(prices[:i+1])

            # 1. Check for GOLDEN RATIO harmonic convergence
            window_start = max(0, i - 20)
            window_prices = prices[window_start:i+1]
            window_prices = window_prices[~np.isnan(window_prices)]

            if len(window_prices) > 5:
                recent_high = np.max(window_prices)
                recent_low = np.min(window_prices)

                if recent_high != recent_low:
                    move = recent_high - recent_low

                    # Sacred Fibonacci levels
                    levels = {
                        'φ²': recent_high - move * (self.phi - 1),  # 0.618
                        'φ': recent_high - move * (2 - self.phi),   # 0.382
                        'φ√3/π': recent_high - move * self.phi_sqrt3_pi  # 0.894
                    }

                    for level_name, level_value in levels.items():
                        # Adaptive threshold based on vortex state
                        if vortex_state['state'] == 'MERKABA_FLOW':
                            threshold = 0.003  # Tight in high flow
                        elif vortex_state['state'] == 'FIRE_TREND':
                            threshold = 0.004  # Medium in trends
                        else:
                            threshold = 0.005  # Standard

                        if abs(current_price - level_value) / current_price < threshold:
                            # Determine signal type based on vortex state
                            if vortex_state['dominant_dr'] == 6:  # Fire trend
                                signal_type = 'BUY' if current_price > np.mean(window_prices) else 'SELL'
                            elif vortex_state['dominant_dr'] == 9:  # Merkaba flow
                                signal_type = 'BUY'  # Always buy in high flow
                            elif vortex_state['dimension'] == 4:  # 4D unity
                                signal_type = 'BALANCED'  # Can go both ways
                            else:
                                signal_type = 'BUY'  # Default

                            # Calculate position parameters
                            strength = 'vortex' if level_name == 'φ√3/π' else 'strong'
                            pos_params = self._calculate_hyper_position(vortex_state, current_price, strength)

                            signals.append({
                                'date': date,
                                'symbol': symbol,
                                'type': signal_type,
                                'price': float(current_price),
                                'reason': f'Sacred {level_name} harmonic [{vortex_state["state"]}]',
                                'strength': strength,
                                'vortex_state': vortex_state.copy(),
                                'target': float(current_price * (1 + pos_params['target_pct'])),
                                'stop': float(current_price * (1 - pos_params['stop_pct'])),
                                'risk_pct': pos_params['risk_pct'],
                                'position_params': pos_params
                            })

            # 2. Check for VORTEX NUMBER price changes
            if i > 0:
                prev_price = prices[i-1]
                if not np.isnan(prev_price):
                    price_change = current_price - prev_price
                    if abs(price_change) > 0.01:  # Minimum change
                        change_in_cents = abs(int(price_change * 100))
                        dr = self._digital_root(change_in_cents)

                        # Only act on sacred numbers
                        if dr in [3, 6, 9]:
                            # Check for vortex sequence alignment
                            seq_index = self.vortex_sequence.index(dr) if dr in self.vortex_sequence else -1
                            if seq_index >= 0:
                                # Next in sequence prediction
                                next_dr = self.vortex_sequence[(seq_index + 1) % len(self.vortex_sequence)]

                                # Calculate position based on vortex progression
                                if dr == 3 and next_dr == 6:
                                    strength = 'medium'  # Triangle to tetrahedron
                                elif dr == 6 and next_dr == 9:
                                    strength = 'strong'  # Tetra to merkaba
                                elif dr == 9 and next_dr == 1:
                                    strength = 'vortex'  # Merkaba to 4D unity
                                else:
                                    strength = 'weak'

                                pos_params = self._calculate_hyper_position(vortex_state, current_price, strength)

                                signals.append({
                                    'date': date,
                                    'symbol': symbol,
                                    'type': 'BUY' if price_change > 0 else 'SELL',
                                    'price': float(current_price),
                                    'reason': f'Vortex {dr}→{next_dr} progression [{vortex_state["solid"]}]',
                                    'strength': strength,
                                    'vortex_state': vortex_state.copy(),
                                    'target': float(current_price * (1 + pos_params['target_pct'])),
                                    'stop': float(current_price * (1 - pos_params['stop_pct'])),
                                    'risk_pct': pos_params['risk_pct'],
                                    'position_params': pos_params
                                })

            # 3. Check for PLATONIC SOLID completion patterns
            # Look for price clusters that form geometric patterns
            if i > 10:
                cluster_prices = prices[i-10:i+1]
                price_range = np.max(cluster_prices) - np.min(cluster_prices)

                # Check if recent moves approximate platonic edge ratios
                if price_range > 0:
                    normalized_moves = np.diff(cluster_prices) / price_range

                    # Count moves near sacred ratios
                    sacred_ratios = [1/self.phi, self.phi - 1, self.phi_sqrt3_pi]
                    sacred_hits = 0

                    for move in normalized_moves:
                        for ratio in sacred_ratios:
                            if abs(abs(move) - ratio) < 0.05:
                                sacred_hits += 1
                                break

                    if sacred_hits >= 3:  # Multiple sacred ratios detected
                        pos_params = self._calculate_hyper_position(vortex_state, current_price, 'strong')

                        signals.append({
                            'date': date,
                            'symbol': symbol,
                            'type': 'BUY',
                            'price': float(current_price),
                            'reason': f'Platonic geometry completion [{vortex_state["solid"]}]',
                            'strength': 'strong',
                            'vortex_state': vortex_state.copy(),
                            'target': float(current_price * (1 + pos_params['target_pct'])),
                            'stop': float(current_price * (1 - pos_params['stop_pct'])),
                            'risk_pct': pos_params['risk_pct'],
                            'position_params': pos_params
                        })

        # Sort by date
        signals.sort(key=lambda x: x['date'])

        print(f"📡 Found {len(signals)} HYPER-DIMENSIONAL vortex signals")

        # Print vortex statistics
        if signals:
            dr_counts = {3: 0, 6: 0, 9: 0, 1: 0, 5: 0}
            for signal in signals:
                dr = signal['vortex_state']['dominant_dr']
                if dr in dr_counts:
                    dr_counts[dr] += 1

            print(f"   Vortex Distribution: {dr_counts}")

            # Check for 4D unity signals
            four_d_signals = [s for s in signals if s['vortex_state']['dimension'] == 4]
            if four_d_signals:
                print(f"   ⚡ 4D UNITY SIGNALS DETECTED: {len(four_d_signals)}")

        return signals

    def run_hyper_simulation(self, symbol: str) -> Dict:
        """Run hyper-dimensional vortex simulation"""

        print(f"\n🌀 RUNNING HYPER-DIMENSIONAL SIMULATION FOR {symbol}")
        print("="*60)

        # Reset
        self.capital = self.initial_capital
        self.positions = {}
        self.trade_history = []
        self.daily_values = []

        # Get data
        data = self.get_2024_data(symbol)
        if data.empty:
            return {"error": "No 2024 data available"}

        self.data = data

        # Find hyper signals
        signals = self.find_hyper_signals(data, symbol)

        # Execute simulation
        prices = data['Close'].values.astype(float)
        dates = data.index

        print(f"\n⚡ EXECUTING HYPER-DIMENSIONAL TRADES...")

        for i, date in enumerate(dates):
            current_price = prices[i]

            if np.isnan(current_price):
                continue

            # Check for signals
            date_signals = [s for s in signals if s['date'] == date]

            for signal in date_signals:
                self._execute_hyper_trade(signal, current_price)

            # Check exits
            self._check_hyper_exits(current_price, date)

            # Record daily value
            self._record_daily_value(current_price, date)

        # Close remaining positions
        if len(prices) > 0:
            final_price = prices[-1]
            if not np.isnan(final_price):
                self._close_all_positions(final_price, dates[-1])

        # Generate report
        report = self._generate_hyper_report(symbol, data)

        return report

    def get_2024_data(self, symbol: str) -> pd.DataFrame:
        """Get 2024 data (same as before)"""

        print(f"\n📅 DOWNLOADING 2024 DATA FOR {symbol}")

        start_2024 = datetime(2024, 1, 1)
        end_2024 = datetime(2024, 12, 31)
        current_date = datetime.now()
        end_date = min(end_2024, current_date)

        print(f"   Period: {start_2024.date()} to {end_date.date()}")

        try:
            data = yf.download(symbol, start=start_2024, end=end_date, progress=False)

            if data.empty:
                print(f"❌ No data for {symbol} in 2024")
                return pd.DataFrame()

            if isinstance(data.columns, pd.MultiIndex):
                data.columns = data.columns.get_level_values(0)

            if 'Close' not in data.columns and 'Adj Close' in data.columns:
                data['Close'] = data['Adj Close']

            if 'Close' in data.columns:
                data['Close'] = pd.to_numeric(data['Close'], errors='coerce')
                data['Close'] = data['Close'].ffill()

                print(f"✅ Got {len(data)} trading days in 2024")

                first_close = data['Close'].iloc[0] if len(data) > 0 else 0
                last_close = data['Close'].iloc[-1] if len(data) > 0 else 0

                if not np.isnan(first_close) and not np.isnan(last_close):
                    print(f"   First close: ${first_close:.2f}")
                    print(f"   Last close: ${last_close:.2f}")

                    total_return = ((last_close / first_close) - 1) * 100
                    print(f"   2024 Total Return: {total_return:.1f}%")

            return data

        except Exception as e:
            print(f"❌ Download error: {str(e)[:100]}...")
            return pd.DataFrame()

    def _execute_hyper_trade(self, signal: Dict, current_price: float):
        """Execute hyper-dimensional trade"""

        symbol = signal['symbol']

        if signal['type'] in ['BUY', 'SELL', 'BALANCED'] and symbol not in self.positions:
            risk_pct = signal['risk_pct']
            risk_amount = self.capital * risk_pct
            stop_distance = abs(current_price - signal['stop'])

            if stop_distance > 0:
                quantity = risk_amount / stop_distance
            else:
                quantity = (self.capital * risk_pct * 10) / current_price

            cost = quantity * current_price

            if cost <= self.capital:
                self.positions[symbol] = {
                    'quantity': quantity,
                    'entry_price': current_price,
                    'entry_date': signal['date'],
                    'target': signal['target'],
                    'stop': signal['stop'],
                    'reason': signal['reason'],
                    'vortex_state': signal['vortex_state'],
                    'risk_pct': risk_pct,
                    'position_params': signal['position_params']
                }
                self.capital -= cost

                vortex_state = signal['vortex_state']
                vortex_str = f"[{vortex_state['state']} | {vortex_state['solid']} | DR:{vortex_state['dominant_dr']}]"

                self.trade_history.append({
                    'date': signal['date'],
                    'action': signal['type'],
                    'price': current_price,
                    'quantity': quantity,
                    'reason': signal['reason'],
                    'vortex_state': vortex_state,
                    'risk_pct': risk_pct,
                    'capital_before': self.capital + cost,
                    'capital_after': self.capital
                })

                date_str = signal['date'].strftime('%Y-%m-%d') if hasattr(signal['date'], 'strftime') else str(signal['date'])[:10]
                print(f"  {date_str} - {signal['type']} @ ${current_price:.2f} {vortex_str}")
                print(f"     Risk: {risk_pct*100:.1f}%, Target: {((signal['target']/current_price)-1)*100:.1f}%")

    def _check_hyper_exits(self, current_price: float, date: datetime):
        """Check for exits with vortex awareness"""

        symbols_to_remove = []

        for symbol, position in self.positions.items():
            # Check vortex-aware exits
            vortex_state = position['vortex_state']

            # In 4D unity or merkaba flow, use wider stops
            if vortex_state['state'] in ['UNITY', 'MERKABA_FLOW']:
                effective_stop = position['stop'] * 0.995  # 0.5% wider
                effective_target = position['target'] * 1.005  # 0.5% wider
            else:
                effective_stop = position['stop']
                effective_target = position['target']

            if current_price <= effective_stop:
                self._close_hyper_position(symbol, current_price, date, "Vortex Stop")
                symbols_to_remove.append(symbol)
            elif current_price >= effective_target:
                self._close_hyper_position(symbol, current_price, date, "Vortex Target")
                symbols_to_remove.append(symbol)

        for symbol in symbols_to_remove:
            del self.positions[symbol]

    def _close_hyper_position(self, symbol: str, price: float, date: datetime, reason: str):
        """Close hyper position"""

        position = self.positions[symbol]
        quantity = position['quantity']
        entry_price = position['entry_price']

        profit_pct = (price - entry_price) / entry_price * 100
        proceeds = quantity * price

        self.capital += proceeds

        vortex_state = position['vortex_state']
        vortex_str = f"[{vortex_state['state']} | DR:{vortex_state['dominant_dr']}]"

        self.trade_history.append({
            'date': date,
            'action': 'CLOSE',
            'price': price,
            'quantity': quantity,
            'reason': f"{reason} {vortex_str}",
            'profit_pct': profit_pct,
            'capital_before': self.capital - proceeds,
            'capital_after': self.capital,
            'vortex_state': vortex_state
        })

        date_str = date.strftime('%Y-%m-%d') if hasattr(date, 'strftime') else str(date)[:10]
        print(f"  {date_str} - CLOSE @ ${price:.2f} - {reason} ({profit_pct:+.1f}%)")

    def _close_all_positions(self, final_price: float, date: datetime):
        """Close all positions"""

        for symbol, position in self.positions.items():
            self._close_hyper_position(symbol, final_price, date, "End of Simulation")

        self.positions.clear()

    def _record_daily_value(self, price: float, date: datetime):
        """Record daily value"""

        portfolio_value = self.capital

        for symbol, position in self.positions.items():
            portfolio_value += position['quantity'] * price

        self.daily_values.append({
            'date': date,
            'value': portfolio_value,
            'cash': self.capital,
            'positions': len(self.positions)
        })

    def _generate_hyper_report(self, symbol: str, data: pd.DataFrame) -> Dict:
        """Generate hyper-dimensional report"""

        if data.empty:
            return {"error": "No data available"}

        if not self.trade_history:
            return {"error": "No trades executed"}

        completed_trades = [t for t in self.trade_history if 'profit_pct' in t]

        if not completed_trades:
            return {"error": "No completed trades"}

        # Overall performance
        first_close = data['Close'].iloc[0] if len(data) > 0 else 0
        last_close = data['Close'].iloc[-1] if len(data) > 0 else 0

        if first_close > 0 and last_close > 0:
            buy_and_hold_return = ((last_close / first_close) - 1) * 100
        else:
            buy_and_hold_return = 0

        vortex_return = ((self.capital / self.initial_capital) - 1) * 100

        # Trade metrics
        profits = [t['profit_pct'] for t in completed_trades]
        winning_trades = [t for t in completed_trades if t['profit_pct'] > 0]
        losing_trades = [t for t in completed_trades if t['profit_pct'] <= 0]

        win_rate = len(winning_trades) / len(completed_trades) * 100 if completed_trades else 0
        avg_win = np.mean([t['profit_pct'] for t in winning_trades]) if winning_trades else 0
        avg_loss = np.mean([t['profit_pct'] for t in losing_trades]) if losing_trades else 0

        total_win = sum([t['profit_pct'] for t in winning_trades])
        total_loss = abs(sum([t['profit_pct'] for t in losing_trades]))
        profit_factor = total_win / total_loss if total_loss > 0 else float('inf')

        # Vortex dimension analysis
        dimension_stats = {}
        vortex_state_stats = {}

        for trade in completed_trades:
            vortex_state = trade.get('vortex_state', {})
            dimension = vortex_state.get('dimension', 3)
            state = vortex_state.get('state', 'UNKNOWN')

            if dimension not in dimension_stats:
                dimension_stats[dimension] = {'count': 0, 'total_pct': 0, 'wins': 0}
            if state not in vortex_state_stats:
                vortex_state_stats[state] = {'count': 0, 'total_pct': 0, 'wins': 0}

            dimension_stats[dimension]['count'] += 1
            dimension_stats[dimension]['total_pct'] += trade['profit_pct']
            if trade['profit_pct'] > 0:
                dimension_stats[dimension]['wins'] += 1

            vortex_state_stats[state]['count'] += 1
            vortex_state_stats[state]['total_pct'] += trade['profit_pct']
            if trade['profit_pct'] > 0:
                vortex_state_stats[state]['wins'] += 1

        # Sharpe ratio
        daily_returns = []
        for i in range(1, len(self.daily_values)):
            prev_val = self.daily_values[i-1]['value']
            curr_val = self.daily_values[i]['value']
            if prev_val > 0:
                ret = (curr_val - prev_val) / prev_val
                daily_returns.append(ret)

        if daily_returns:
            sharpe = np.mean(daily_returns) / np.std(daily_returns) * np.sqrt(252) if np.std(daily_returns) > 0 else 0
        else:
            sharpe = 0

        # Max drawdown
        values = [v['value'] for v in self.daily_values]
        if values:
            peak = values[0]
            max_dd = 0

            for value in values:
                if value > peak:
                    peak = value
                dd = (peak - value) / peak * 100 if peak > 0 else 0
                if dd > max_dd:
                    max_dd = dd
        else:
            max_dd = 0

        # Prepare samples
        trade_samples = []
        for trade in completed_trades[-10:]:
            trade_copy = trade.copy()
            if isinstance(trade_copy['date'], pd.Timestamp):
                trade_copy['date'] = trade_copy['date'].strftime('%Y-%m-%d')
            elif hasattr(trade_copy['date'], 'strftime'):
                trade_copy['date'] = trade_copy['date'].strftime('%Y-%m-%d')
            trade_samples.append(trade_copy)

        report = {
            'symbol': symbol,
            'year': 2024,
            'trading_days': len(data),
            'market_return_pct': float(buy_and_hold_return),
            'vortex_return_pct': float(vortex_return),
            'outperformance_pct': float(vortex_return - buy_and_hold_return),
            'initial_capital': float(self.initial_capital),
            'final_capital': float(self.capital),
            'total_trades': len(completed_trades),
            'winning_trades': len(winning_trades),
            'losing_trades': len(losing_trades),
            'win_rate_pct': float(win_rate),
            'avg_win_pct': float(avg_win),
            'avg_loss_pct': float(avg_loss),
            'profit_factor': float(profit_factor),
            'sharpe_ratio': float(sharpe),
            'max_drawdown_pct': float(max_dd),
            'best_trade_pct': float(max(profits)) if profits else 0,
            'worst_trade_pct': float(min(profits)) if profits else 0,
            'dimension_stats': dimension_stats,
            'vortex_state_stats': vortex_state_stats,
            'trade_samples': trade_samples,
            'summary': self._get_hyper_summary(vortex_return, win_rate, profit_factor)
        }

        return report

    def _get_hyper_summary(self, return_pct: float, win_rate: float, profit_factor: float) -> str:
        """Get hyper summary"""

        if return_pct > 100:
            assessment = "🌀 HYPER-DIMENSIONAL BREAKTHROUGH"
        elif return_pct > 50:
            assessment = "🔥 VORTEX ASCENSION ACHIEVED"
        elif return_pct > 30:
            assessment = "✅ SACRED GEOMETRY ALIGNED"
        elif return_pct > 20:
            assessment = "👍 VORTEX HARMONY ACTIVE"
        elif return_pct > 10:
            assessment = "⚠️ GEOMETRIC RESONANCE WEAK"
        elif return_pct > 0:
            assessment = "📉 MINIMAL VORTEX FLOW"
        else:
            assessment = "❌ VORTEX COLLAPSE"

        if win_rate > 70:
            assessment += " | High Geometric Precision"
        elif win_rate < 40:
            assessment += " | Chaotic Patterns"

        if profit_factor > 3:
            assessment += " | Perfect Ratio Balance"
        elif profit_factor < 1.5:
            assessment += " | Ratio Discord"

        return assessment

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def run_hyper_vortex_test():
    """Run hyper-dimensional vortex test"""

    print("🚀 HYPER-DIMENSIONAL 369 VORTEX SYSTEM")
    print("="*70)
    print("Integrating: Platonic Solids • 4D Geometry • Sacred Ratios • Vortex Math")
    print("Sequence: 3(triangle) → 6(tetra) → 9(merkaba) → 1(4D unity) → 5(hyper) → 3")
    print("Sacred Ratio: φ√3/π = {:.3f}".format((1 + np.sqrt(5)) / 2 * np.sqrt(3) / np.pi))
    print("="*70)

    symbols = ["SPY", "QQQ", "BTC-USD", "TSLA", "NVDA"]

    all_results = {}

    for symbol in symbols:
        print(f"\n" + "="*70)
        print(f"🌀 TESTING HYPER-DIMENSIONAL VORTEX: {symbol}")
        print("="*70)

        trader = HyperVortexSystem(initial_capital=10000)
        report = trader.run_hyper_simulation(symbol)

        if 'error' not in report:
            # Print summary
            print(f"\n📊 HYPER-VORTEX PERFORMANCE:")
            print(f"   Market Return: {report['market_return_pct']:.1f}%")
            print(f"   Vortex Return: {report['vortex_return_pct']:.1f}%")
            print(f"   Outperformance: {report['outperformance_pct']:+.1f}%")
            print(f"   Win Rate: {report['win_rate_pct']:.1f}%")
            print(f"   Trades: {report['total_trades']}")

            # Dimension analysis
            if 'dimension_stats' in report:
                print(f"\n📈 DIMENSION ANALYSIS:")
                for dim, stats in report['dimension_stats'].items():
                    win_rate = stats['wins'] / stats['count'] * 100 if stats['count'] > 0 else 0
                    avg = stats['total_pct'] / stats['count'] if stats['count'] > 0 else 0
                    print(f"   {dim}D: {stats['count']} trades, {win_rate:.1f}% WR, Avg: {avg:+.1f}%")

            # Vortex state analysis
            if 'vortex_state_stats' in report:
                print(f"\n🌀 VORTEX STATE PERFORMANCE:")
                for state, stats in report['vortex_state_stats'].items():
                    if stats['count'] > 0:
                        win_rate = stats['wins'] / stats['count'] * 100
                        avg = stats['total_pct'] / stats['count']
                        print(f"   {state}: {stats['count']} trades, {win_rate:.1f}% WR, Avg: {avg:+.1f}%")

            print(f"\n💎 SUMMARY: {report['summary']}")

            all_results[symbol] = {
                'market_return': report['market_return_pct'],
                'vortex_return': report['vortex_return_pct'],
                'outperformance': report['outperformance_pct'],
                'win_rate': report['win_rate_pct'],
                'profit_factor': report['profit_factor']
            }
        else:
            print(f"❌ Skipping {symbol}: {report['error']}")

    # Final summary
    print("\n" + "="*70)
    print("📊 HYPER-DIMENSIONAL VORTEX - FINAL REPORT")
    print("="*70)

    if all_results:
        print(f"\n{'Symbol':<10} {'Market %':<10} {'Vortex %':<10} {'OutPerf %':<12} {'Win Rate':<10} {'Profit Factor':<12}")
        print("-" * 70)

        for symbol, results in all_results.items():
            market = f"{results['market_return']:.1f}%"
            vortex = f"{results['vortex_return']:.1f}%"
            outperf = f"{results['outperformance']:+.1f}%"
            win_rate = f"{results['win_rate']:.1f}%"
            pf = f"{results['profit_factor']:.2f}"

            print(f"{symbol:<10} {market:<10} {vortex:<10} {outperf:<12} {win_rate:<10} {pf:<12}")

        # Averages
        avg_outperformance = np.mean([r['outperformance'] for r in all_results.values()])
        avg_win_rate = np.mean([r['win_rate'] for r in all_results.values()])

        print("\n📈 HYPER-DIMENSIONAL AVERAGES:")
        print(f"   Average Outperformance: {avg_outperformance:+.1f}%")
        print(f"   Average Win Rate: {avg_win_rate:.1f}%")

        # Vortex verdict
        print("\n" + "="*70)
        if avg_outperformance > 20:
            print("🌀 HYPER-DIMENSIONAL BREAKTHROUGH ACHIEVED")
            print("   4D geometry + vortex math = UNIVERSE-LEVEL EDGE")
        elif avg_outperformance > 0:
            print("✅ SACRED GEOMETRY VALIDATED")
            print("   Vortex mathematics reveals market harmonics")
        else:
            print("❌ VORTEX COLLAPSE DETECTED")
            print("   Even hyper-dimensional math couldn't beat 2024 bull")
        print("="*70)

# ============================================================================
# RUN HYPER VORTEX
# ============================================================================

if __name__ == "__main__":
    run_hyper_vortex_test()
```
